```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sbn

"""
Goal of this notebook is to find features in the dataframe that may be good predictors 
of the final exam score. This will be done with this process:

EDA-> Preprocess Dataset -> Modeling -> Evaluation

"""
```

```python
df = pd.read_csv('student_performance_dataset.csv')
## determine all the column names and dtypes
print(f'columns and dtypes: {df.dtypes} ')
## find missing values
print(f'missing values from each column: {df.isnull().sum()}')
## find duplicatesabs
print(f'duplicated rows: {df.duplicated().sum()}')
```

```python
plt.figure(figsize=(12,12))
corr_matrix = df.corr(numeric_only=True)
sbn.heatmap(corr_matrix,cmap='coolwarm',fmt='.1f',annot=True)
plt.show()
```

```python
## columns with relatively high correlations
hi_corr = ['study_hours_per_day','deep_work_sessions',
           'assignment_completion_rate','social_media_hours',
           'doomscrolling_before_sleep','stress_level',
           'motivation_level','focus_score',
           'procrastination_index','sleep_hours',
           'financial_stress','revision_efficiency',
           'burnout_risk','consistency_score','final_exam_score']
```

```python
## check distribution of final exam scores to verify its approximately normal
print(f"skew: {df['final_exam_score'].skew()}")
print(f"kurtosis: {df['final_exam_score'].kurtosis()}")
print(df['final_exam_score'].describe())

# plot histogram
sbn.histplot(data=df, x='final_exam_score', kde=True, bins=40)
plt.show()
```
